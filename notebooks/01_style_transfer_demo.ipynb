# 🎨 Style Transfer Demo
Demonstrates VGG19-based neural style transfer using the PMG-AI pipeline.

```python
import sys
sys.path.append('../src/pmg1/style_transfer')
from style_transfer import perform_style_transfer

# Set paths to your images
content_image = 'path/to/content.jpg'
style_image = 'path/to/style.jpg'

result = perform_style_transfer(content_image, style_image)
print(f'Stylized image saved to: {result}')
```

```python
from PIL import Image
import matplotlib.pyplot as plt

fig, axes = plt.subplots(1, 3, figsize=(15, 5))
axes[0].imshow(Image.open(content_image)); axes[0].set_title('Content Image'); axes[0].axis('off')
axes[1].imshow(Image.open(style_image)); axes[1].set_title('Style Image'); axes[1].axis('off')
axes[2].imshow(Image.open(result)); axes[2].set_title('Stylized Output'); axes[2].axis('off')
plt.tight_layout(); plt.show()
```
