# 🧊 2D to 3D Reconstruction Demo
Demonstrates depth estimation with MiDaS and 3D point cloud generation using Open3D.

```python
import sys
sys.path.append('../src/pmg2/reconstruction_2d_3d')
from midas_depth import estimate_depth_midas
from depth_to_3d import depth_to_3d
import cv2, numpy as np, open3d as o3d

image_path = 'path/to/input.jpg'
depth = estimate_depth_midas(image_path, output_path='depth_output.jpg')
print('Depth map generated.')
```

```python
import matplotlib.pyplot as plt

depth_vis = cv2.imread('depth_output.jpg', cv2.IMREAD_GRAYSCALE)
plt.figure(figsize=(10, 5))
plt.subplot(1,2,1); plt.imshow(cv2.cvtColor(cv2.imread(image_path), cv2.COLOR_BGR2RGB)); plt.title('Input'); plt.axis('off')
plt.subplot(1,2,2); plt.imshow(depth_vis, cmap='plasma'); plt.title('Depth Map'); plt.axis('off')
plt.show()

pcd = depth_to_3d(depth)
o3d.io.write_point_cloud('output.ply', pcd)
print('Point cloud saved to output.ply')
```
