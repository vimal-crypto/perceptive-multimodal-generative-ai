# 🖼️ Text-to-Image Generation Demo
Demonstrates sentiment-aware image generation using Stable Diffusion from PMG-2.

```python
import sys
sys.path.append('../src/pmg2/text_to_image')
from fgen import generate_image_from_text, analyze_sentiment

prompt = 'A futuristic city with flying cars at golden hour'
print(f'Detected sentiment: {analyze_sentiment(prompt)}')
output = generate_image_from_text(prompt, output_path='generated.png')
print(f'Image saved to: {output}')
```

```python
from PIL import Image
import matplotlib.pyplot as plt

img = Image.open('generated.png')
plt.figure(figsize=(8, 8))
plt.imshow(img)
plt.axis('off')
plt.title(f'Generated: {prompt}')
plt.show()
```
